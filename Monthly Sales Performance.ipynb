```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a quick data dictionary (Fixed with actual data!)
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [10, 15, 25, 20, 30]
}

# 2. Load it into a Pandas DataFrame
df = pd.DataFrame(data)

# 3. Use our indentation rules to scan the data
for index, row in df.iterrows():
    if row["Sales"] > 15:
        # Level 3 indentation (8 spaces)
        print(f"🔥 High sales month detected: {row['Month']}")

# 4. Plot the data instantly below the cell
df.plot(x="Month", y="Sales", kind="line", marker="o", color="blue")
plt.title("Monthly Sales Performance")
plt.show()
```
